```python
import torch
```

#### Define netural network layers 

```python
W1 = torch.tensor([[1.,1.], [1.,1.]], requires_grad = True)
W2 = torch.tensor([1., -2.], requires_grad = True)
b12 = torch.tensor([-0.5, 1.5], requires_grad = True)
b3 = torch.tensor(-0.5, requires_grad = True)
x = torch.tensor([1., 0.])
```

#### Define neural network computations

```python
NN = lambda x : torch.sigmoid(torch.matmul(W2,torch.sigmoid(torch.matmul(W1,x)+b12))+b3)
```

#### Feed x to NN

```python
NN(x)
```

#### Run backpropagation using PyTorch's backward() function

```python
NN(x).backward()
```
