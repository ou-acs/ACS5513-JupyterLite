### Autodiff

Calculate the gradient of the function $f(x) = \ln(\exp(3x) -\cos(x))$ at $x=.5$ using autodiff.

```python
 import torch
```

```python
x = torch.tensor(0.5, requires_grad=True)
p = 3*x
q = torch.exp(p)
r = torch.cos(x)
s = q - r
f = torch.log(s)
print(f)
```

```python
f.backward()
print(x.grad)
```
