### Building a Single Layer Perceptron

```python
import numpy as np
```

```python
class Perceptron:
    def __init__(self, eta = 0.01, epochs=100):
        self.eta = eta
        self.epochs = epochs
        self.weights = None
        self.bias = None

    def activation(self, z):
        return np.where(z >= 0, 1, 0)  #Heaviside function
        
    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0
        for i in range(self.epochs):
            print(f"Epoch:{i}, weights: {self.weights}, bias: {self.bias}")
            z = X @ self.weights + self.bias
            y_pred = self.activation(z)
            error = y - y_pred
            self.weights += self.eta * (X.T @ error)
            self.bias += self.eta * np.sum(error)
                
    def predict(self, X):
        z = X @ self.weights + self.bias
        return self.activation(z)
                
        
```

#### Train the Perceptron and analyze its changing weights and bias in each epoch. 

```python
X = np.array([[0,0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 0, 0, 1])

model = Perceptron(eta = 0.1, epochs = 5)
model.fit(X, y)
```

#### Predict y for X. Change eta to different values and analyze how the predictions change. 

```python
predictions = model.predict(X)
print(predictions)
```
