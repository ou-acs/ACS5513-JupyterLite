```python
import torch
```

```python
x = torch.tensor(torch.pi/2, requires_grad=True)
```

```python
print(x)
```

#### Take function $f(x) = sin(x) + x^2$

```python
f = torch.sin(x) + x**2
```

#### Take f's gradient in backpropagation

```python
f.backward()
print(x.grad)
```

#### IMPORTANT
If we want to compute the gradient of x again, we need to reset its value as otherwise the gradient is accumulated. 

##### For example, without resetting:

```python
f = torch.sin(x) + x**2
f.backward()
print(x.grad)
```

##### Resetting

```python
x.grad.zero_()
f = torch.sin(x) + x**2
f.backward()
print(x.grad)
```
