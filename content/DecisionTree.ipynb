### Decision Tree

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

#### Define a pre-trained decision tree

##### Create a Node class that will be used to build the tree.

```python
class Node:
    def __init__(self):
        self.left = None
        self.right = None
        self.key = None
        self.val = None
```

##### Use instances of Node class to create the decision tree starting with the root, and linking child nodes and leaves

```python
# root node
root = Node()
root.key = 'x1'
root.val = 0.5

#left node
x2 = Node()
x2.key = 'x2'
x2.val = 0.5
x2.left = Node()
x2.left.val = 1
x2.right = Node()
x2.right.val = 0
root.left = x2

#right node
x3 = Node()
x3.key = 'x3'
x3.val = 0.5

#right -> left node
x4 = Node()
x4.key = 'x4'
x4.val = 0.5
x4.left = Node()
x4.left.val = 0
x4.right = Node()
x4.right.val = 1
x3.left = x4

x3.right = Node()
x3.right.val = 0

root.right = x3
```

##### Create the classifier with predict function that takes the root of the decision tree and the feature-values for which the label needs to be predicted. 

```python
class DecisionTreeClassifier:
    def __init__(self, root):
        self.root = root
        
    def predict(self, node, features):
        current_node = node
        while (True):
            if current_node.left is None and current_node.right is None:
                return current_node.val
            elif features[current_node.key] < current_node.val:
                current_node = current_node.left
            else:
                current_node = current_node.right
        
```

##### Use a test sample and use DecisionTreeClassifier to predict its label. 

```python
sample = {
    'x1': 0.6,
    'x2': 0.2,
    'x3': 0.2,
    'x4': 0.6
}
```

```python
dtc = DecisionTreeClassifier(root)
print(dtc.predict(root, sample))
```
